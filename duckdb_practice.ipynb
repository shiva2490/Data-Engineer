```python
import pandas as pd
import duckdb

# 1. Create a standard Pandas DataFrame
df_sales = pd.DataFrame({
    'Team': ['Data_Engineering', 'Data_Engineering', 'AI_Research', 'AI_Research'],
    'Project': ['Pipeline A', 'Database B', 'Model X', 'Vision Y'],
    'Score': [85, 95, 88, 92]
})

print("--- Original Pandas Data ---")
display(df_sales)
```

```python
# 2. Write your SQL Query as a text string
# Notice how we just select FROM 'df_sales' (the Python variable name!)

my_query = """
SELECT
    Team,
    AVG(Score) AS Average_Score
FROM df_sales
GROUP BY TEAM;
"""

# 3. Run the SQL query using DuckDB and output the result as a new Pandas table (.df)
sql_result = duckdb.query(my_query).df()

print("--- DuckDB SQL Result ---")
display(sql_result)
```

```python
# Write a SQL query to find the total revenue and total quantity sold for each product in the 'Electronics' category, 
# but only include products where the total revenue generated is greater than $500.
import pandas as pd
import duckdb

df_products = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106],
    'Category': ['Electronics', 'Electronics', 'Clothing', 'Electronics', 'Clothing', 'Electronics'],
    'Product': ['Laptop', 'Mouse', 'T-Shirt', 'Laptop', 'Jeans', 'Mouse'],
    'Quantity': [1, 5, 2, 2, 1, 10],
    'Price': [1200.00, 50.00, 25.00, 1200.00, 40.00, 50.00]
})


df_products
```

```python
query_1 = """
SELECT 
    Product,
    SUM(Quantity) AS Total_Quantity,
    SUM(Quantity * price) As Total_Revenue
FROM df_products
WHERE Category = 'Electronics'
GROUP BY Product
Having SUM(Quantity * price) > 500;
"""

duckdb.query(query_1).df()
```

```python
import pandas as pd
import duckdb

df_hr = pd.DataFrame({
    'EmpName': ['Amit', 'Priya', 'Suresh', 'Rahul', 'Deepa', 'Ananya'],
    'Department': ['IT', 'IT', 'HR', 'Marketing', 'Marketing', 'IT'],
    'JobTitle': ['Engineer', 'Manager', 'Recruiter', 'Analyst', 'Manager', 'Engineer'],
    'Salary': [80000, 120000, 60000, 75000, 110000, 85000]
})

df_hr
```

```python
# Your Task: Write a query that groups the employees by Department. For each department, 
# calculate:
# The Total Budget (the sum of all salaries).
# The Employee Count (how many people work there). Hint: Use COUNT(EmpName).
# The Filter: Only show departments where the Total Budget is greater than $150,000.
query_2 = """
SELECT
    Department,
    SUM(Salary) AS Total_Budget,
    COUNT(EmpName) AS Employee_Count
FROM df_hr
GROUP BY Department
Having SUM(Salary) > 150000;
"""
duckdb.query(query_2).df()
```

```python
import pandas as pd
df_logistics = pd.DataFrame({
    'ShipmentID': [1, 2, 3, 4, 5, 6, 7],
    'Warehouse': ['Mumbai', 'Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Chennai'],
    'TransportMode': ['Air', 'Road', 'Air', 'Air', 'Air', 'Road', 'Air'],
    'Weight_kg': [300, 800, 250, 600, 150, 400, 100]
})

df_logistics
```

```python
# Your Task: Write a query to find the Average Weight and Total Weight of packages sent from each Warehouse.
# The Rules:
# First, only look at packages where TransportMode is exactly 'Air'.
# Group the data by Warehouse.
# Finally, only show the warehouses where their Total Weight of Air packages is greater than 500 kg.

import duckdb
query_3 = """
SELECT
    Warehouse,
    SUM(Weight_kg) AS Total_weight,
    AVG(Weight_kg) AS Average_weight
FROM df_logistics
WHERE TransportMode = 'Air'
GROUP BY Warehouse
Having SUM(Weight_kg) > 500
"""

duckdb.query(query_3).df()
```
