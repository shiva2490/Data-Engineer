```python
# Joins
import pandas as pd

# Table 1: Passenger Names
df_names = pd.DataFrame({
    'PassengerId': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})
print(df_names)

# Table 2: Ticket Prices
df_prices = pd.DataFrame({
    'PassengerId': [1, 2, 4],
    'Fare': [50.00, 80.00, 100.00]
})

# here display function display same like a table structure
display(df_prices)
```

```python
df_names
```

```python
df_prices
```

```python
# Joining tables
# inner joins (display only common data) similar to intersection in sets
pd.merge(df_names, df_prices, on='PassengerId', how='inner')
```

```python
#Left Join display only left side table data(if ids match data display not displayed null(NAN) values)
pd.merge(df_names, df_prices, on='PassengerId', how='left')
```

```python
# Right join(it will disaply right side table if ids match data displayed if not NAN values)
pd.merge(df_names, df_prices, on='PassengerId', how='right')
```

```python
# Advanced Transformations
# 1. Data Type Casting (.astype()) -> 'float', 'int', 'str'
# 2. Text Cleaning (.str accessor) -> .str.strip() to remove blank spaces at beginning or end, .str.lower(), .str.upper()

import pandas as pd

df_messy = pd.DataFrame({
     'Passenger': [' Alice ', 'Bob', 'Charlie '],
     'TicketCode': [101, 102, 103]
})

df_messy
```

```python
# 1. Fix the TicketCode (Data Type Casting) int to str

df_messy['TicketCode'] = df_messy['TicketCode'].astype('str')
df_messy
```

```python
# 2. Clean the Passenger Names (Text Cleaning)
df_messy['Passenger'] = df_messy['Passenger'].str.strip()
```

```python
# Setup our family dataset
family_data = {
    'Passenger': ['Alice', 'Bob', 'Charlie'],
    'SibSp': [1, 0, 2], # Number of Siblings or Spouses aboard
    'Parch': [1, 0, 1]  # Number of Parents or Children aboard
}
df_family = pd.DataFrame(family_data)
df_family
```

```python
# 1.Your manager wants to know the Total Family Size for each passenger. Total family size is calculated by adding the SibSp column to the Parch column.
df_family['TotalFamily'] = df_family['SibSp'] + df_family['Parch']
```

```python
df_family
```

```python
# 2: Aggregations (.groupby)

# Setup our aggregation dataset
agg_data = {
    'Pclass': [1, 1, 2, 2, 3, 3], # Passenger Class
    'Fare': [100.0, 150.0, 50.0, 60.0, 10.0, 15.0] # Ticket Price
}
df_agg = pd.DataFrame(agg_data)
df_agg
```

```python
# What was the average ticket fare for each passenger class?
df_agg.groupby('Pclass')['Fare'].mean()
```
