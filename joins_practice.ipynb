```python
# Joins
import pandas as pd

# Table 1: Passenger Names
df_names = pd.DataFrame({
    'PassengerId': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})
print(df_names)

# Table 2: Ticket Prices
df_prices = pd.DataFrame({
    'PassengerId': [1, 2, 4],
    'Fare': [50.00, 80.00, 100.00]
})

# here display function display same like a table structure
display(df_prices)
```

```python
df_names
```

```python
df_prices
```

```python
# Joining tables
# inner joins (display only common data) similar to intersection in sets
pd.merge(df_names, df_prices, on='PassengerId', how='inner')
```

```python
#Left Join display only left side table data(if ids match data display not displayed null(NAN) values)
pd.merge(df_names, df_prices, on='PassengerId', how='left')
```

```python
# Right join(it will disaply right side table if ids match data displayed if not NAN values)
pd.merge(df_names, df_prices, on='PassengerId', how='right')
```
