```python
import pandas as pd
import duckdb

df_sales = pd.DataFrame({
    'RepName': ['John', 'Alice', 'Bob', 'Emma', 'Liam'],
    'Region': ['North', 'North', 'North', 'South', 'South'],
    'Sales': [10000, 15000, 10000, 20000, 18000]
})

df_sales
```

```python
#Write a SQL query that displays the RepName, Region, Sales, and a new calculated column called Regional_Rank.
#Use DENSE_RANK() to calculate the rank.
#The rank should reset for every new Region.
#The person with the highest sales in that region should get Rank 1.

query_C = """
SELECT 
    RepName,
    Region,
    Sales,
    DENSE_RANK() OVER (PARTITION BY Region ORDER BY Sales DESC) AS Regional_Rank
FROM df_sales;
"""

duckdb.query(query_C).df()
# PARTITION BY means GROUP BY
```

```python
# Your Task: 1. Filter the data so you are only looking at sensors where the Status is exactly 'Active'.
# 2. Create a Temperature_Rank column using a Window Function. Rank the temperatures from highest to lowest, but restart the ranking for each individual City.

import pandas as pd

df_sensors = pd.DataFrame({
    'City': ['Tokyo', 'Kyoto', 'Tokyo', 'Tokyo', 'Kyoto', 'Kyoto'],
    'SensorID': ['T-1', 'K-1', 'T-2', 'T-3', 'K-2', 'K-3'],
    'Status': ['Active', 'Active', 'Offline', 'Active', 'Active', 'Offline'],
    'Temperature': [39.0, 35.0, 99.9, 37.2, 38.5, -99.9]
})

df_sensors
```

```python
import duckdb

query_1 = """
SELECT
    City,
    SensorID,
    Temperature,
    DENSE_RANK() OVER(PARTITION BY City ORDER BY Temperature DESC) AS Temperature_Rank
FROM df_sensors
WHERE Status = 'Active';
"""
duckdb.query(query_1).df()
```

```python
import pandas as pd
import duckdb

df_server = pd.DataFrame({
    'Date': ['2026-06-01', '2026-06-02', '2026-06-03', '2026-06-04'],
    'Traffic': [500, 550, 800, 750]
})

query_F = """
SELECT 
    Date,
    Traffic,
    LAG(Traffic) OVER (ORDER BY Date) AS Previous_Traffic,
    Traffic - LAG(Traffic) OVER (ORDER BY Date) AS Traffic_Spike
FROM df_server;
"""

duckdb.query(query_F).df()
```

```python
import pandas as pd

df_clicks = pd.DataFrame({
    'UserID': ['U-101', 'U-101', 'U-101', 'U-102', 'U-102'],
    'Timestamp': ['09:00', '09:05', '09:10', '10:15', '10:20'],
    'Current_Page': ['/home', '/pricing', '/checkout', '/home', '/docs']
})
df_clicks
```

```python
# Write a query to calculate the Next_Page.
# Use LEAD() to look forward.
# Use PARTITION BY so you don't accidentally mix up User 1's journey with User 2's journey!
# Use ORDER BY to make sure time is moving forward chronologically.

import duckdb
query_nextpage = """
SELECT
    Userid,
    timestamp,
    current_page,
    Lead(current_page) OVER(partition by userid order by timestamp asc) AS Next_Page
FROM df_clicks;
"""

duckdb.query(query_nextpage).df()
```

```python
import pandas as pd

df_performance = pd.DataFrame({
    'Server_ID': ['Server_1', 'Server_1', 'Server_1', 'Server_2', 'Server_2'],
    'Timestamp': ['08:00', '08:05', '08:10', '09:00', '09:15'],
    'Latency_ms': [250, 45, 50, 300, 60]
})

df_performance
```

```python
# The database groups all the logs for Server A.
# It sorts them chronologically.
# It looks at the very first row, sees a latency of 150ms, and permanently holds that number in its memory.
# As it walks down the rest of the rows for Server A, it pastes 150ms next to every single one.
import duckdb
query_latency = """
SELECT
    server_id,
    timestamp,
    latency_ms,
    FIRST_VALUE(latency_ms) OVER(PARTITION BY SERVER_ID ORDER BY TIMESTAMP ASC) AS Baseline_Latency
FROM df_performance;
"""
duckdb.query(query_latency).df()
```

# NTILE()
#### Imagine you have a list of people sorted by their test scores.Instead of giving them individual rank numbers (1st place, 2nd place), NTILE(x) chops that list into x equal-sized buckets.
#### In the real world, the Marketing or Sales teams don't want to look at 10,000 individual users. They want to group people into tiers so they can take action.
#### 1. "Give me the Top 10% of spenders so we can send them a VIP gift." -> NTILE(10)
#### 2. "Divide our servers into 3 tiers (Fast, Normal, Slow)." -> NTILE(3)
#### 3. "Split our students into Quartiles (Top 25%, etc.)." -> NTILE(4)

```python
import pandas as pd

df_customers = pd.DataFrame({
    'Customer_Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Fiona'],
    'Total_Spend': [1000, 300, 100, 800, 500, 50]
})
df_customers
```

```python
# You are preparing a dataset for the marketing team. They want to send out three different email campaigns: a VIP promo, a Standard promo, 
# and a Re-engagement promo. They need you to divide all the customers into exactly 3 tiers based on their total lifetime spending.

import duckdb

query_ntile = """
SELECT
    Customer_Name,
    Total_Spend,
    NTILE(3) OVER(ORDER BY total_spend DESC) AS Customer_Tier
FROM df_customers;
"""
duckdb.query(query_ntile).df()
```

# LAST_VALUE()
#### similar to FIRST_VALUE() but just opposite

```python
import pandas as pd

df_routes = pd.DataFrame({
    'Driver': ['Driver_A', 'Driver_A', 'Driver_A', 'Driver_B', 'Driver_B'],
    'Time': ['08:00', '08:30', '09:00', '10:00', '11:00'],
    'Stop_Name': ['Warehouse', 'Coffee Shop', 'Post Office', 'Warehouse', 'Gas Station']
})

df_routes
```

```python
# You are analyzing delivery driver routes. The Logistics Manager wants a report that shows every stop a driver made, 
# but they also want a column showing the driver's Final Destination on that exact same row.

import duckdb
query_fd = """
SELECT
    Driver,
    Time,
    Stop_Name,
    LAST_VALUE(Stop_Name) OVER(
        PARTITION BY Driver
        ORDER BY Time ASC
        ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING
        ) AS Final_Destination
FROM df_routes;
"""

duckdb.query(query_fd).df()
```

```python
# You are analyzing delivery driver routes. The Logistics Manager wants a report that shows every stop a driver made, 
# but they also want a column showing the driver's Final Destination on that exact same row.

import duckdb
query_fd = """
SELECT
    Driver,
    Time,
    Stop_Name,
    LAST_VALUE(Stop_Name) OVER(
        PARTITION BY Driver
        ORDER BY Time ASC
        ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING
        ) AS Final_Destination
FROM df_routes
ORDER BY Driver, Time ASC;
"""

duckdb.query(query_fd).df()
```

```python
import pandas as pd

df_sales = pd.DataFrame({
    'Emp_Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Revenue': [50000, 80000, 30000, 120000, 95000, 45000, 110000, 60000]
})

df_sales
```

```python
# The VP of Sales is calculating end-of-year bonuses. They hand you a dataset of every salesperson and their total revenue.
# They have two specific requests for the final report:
# 1. The Tiers: They want to divide the team into exactly 4 equal groups (Quartiles) based on revenue. The highest earners must be in Tier 1.
# 2. The Target to Beat: To motivate the team, they want a column that displays the revenue of the absolute #1 Top Performer pasted onto every single row, so everyone knows exactly what the high score was.

import duckdb
query_z = """
SELECT
    Emp_Name,
    Revenue,
    NTILE(4) OVER(ORDER BY Revenue DESC) AS Performance_Tier,
    FIRST_VALUE(Emp_Name) OVER(ORDER BY Revenue DESC) AS Top_Scorer
FROM df_sales;
"""
duckdb.query(query_z).df()
```
